```python
import os
from collections import Counter

dataset_folder = "dataset"

extensions = Counter()

for root, dirs, files in os.walk(dataset_folder):
    for file in files:
        ext = os.path.splitext(file)[1].lower()
        if ext:
            extensions[ext] += 1

print("File Extension Summary")
print("======================")

for ext, count in extensions.items():
    print(f"{ext}: {count}")

print("\nTotal files:", sum(extensions.values()))
```

```python
import os
from collections import Counter

dataset_folder = "dataset"

image_extensions = [".jpg", ".jpeg", ".png", ".bmp", ".webp"]

image_files = []

for root, dirs, files in os.walk(dataset_folder):
    for file in files:
        if os.path.splitext(file)[1].lower() in image_extensions:
            image_files.append(os.path.join(root, file))

print("===================================")
print("DATASET SUMMARY")
print("===================================")

print("Total images:", len(image_files))

# Count images by extension
extension_count = Counter(
    os.path.splitext(file)[1].lower()
    for file in image_files
)

print("\nImage types:")
for ext, count in extension_count.items():
    print(f"{ext}: {count}")

# Count images by folder
folder_count = Counter(
    os.path.dirname(file)
    for file in image_files
)

print("\nImages in each folder:")
for folder, count in folder_count.items():
    print(f"{folder}: {count}")
```

```python
from PIL import Image
import os
from collections import Counter

dataset_folder = "dataset"

image_extensions = [".jpg", ".jpeg", ".png", ".bmp", ".webp"]

image_files = []

for root, dirs, files in os.walk(dataset_folder):
    for file in files:
        if os.path.splitext(file)[1].lower() in image_extensions:
            image_files.append(os.path.join(root, file))

print("===================================")
print("IMAGE DIMENSION & QUALITY CHECK")
print("===================================")

sizes = Counter()
valid_images = 0
corrupted_images = []

for file in image_files:
    try:
        with Image.open(file) as img:
            sizes[img.size] += 1
            valid_images += 1
    except Exception:
        corrupted_images.append(file)

print("Total image files:", len(image_files))
print("Valid images:", valid_images)
print("Corrupted/unreadable images:", len(corrupted_images))

print("\nMost common image dimensions:")
for size, count in sizes.most_common(10):
    print(f"{size}: {count} images")

if corrupted_images:
    print("\nCorrupted image examples:")
    for file in corrupted_images[:10]:
        print("-", file)
else:
    print("\n✅ No corrupted images detected.")
```

```python
import os
from collections import Counter

dataset_folder = "dataset"

image_extensions = [".jpg", ".jpeg", ".png", ".bmp", ".webp"]

# Find all image files
image_files = []

for root, dirs, files in os.walk(dataset_folder):
    for file in files:
        if os.path.splitext(file)[1].lower() in image_extensions:
            image_files.append(os.path.join(root, file))

# Use the immediate parent folder as the class/category
classes = []

for file in image_files:
    class_name = os.path.basename(os.path.dirname(file))
    classes.append(class_name)

class_counts = Counter(classes)

print("===================================")
print("CLASS / CATEGORY DISTRIBUTION")
print("===================================")

print("Total images:", len(image_files))
print("Number of classes/categories:", len(class_counts))

print("\nImages per class:")
for class_name, count in class_counts.items():
    print(f"{class_name}: {count}")
```

```python
from PIL import Image
import os

print("===================================")
print("MISSING / INVALID IMAGE CHECK")
print("===================================")

missing_files = []
invalid_images = []
zero_size_images = []

for file in image_files:

    # Check if file exists
    if not os.path.exists(file):
        missing_files.append(file)
        continue

    # Check if file size is zero
    if os.path.getsize(file) == 0:
        zero_size_images.append(file)
        continue

    # Check if image can be opened
    try:
        with Image.open(file) as img:
            img.verify()
    except Exception:
        invalid_images.append(file)

print("Total images checked:", len(image_files))
print("Missing files:", len(missing_files))
print("Zero-size files:", len(zero_size_images))
print("Invalid/corrupted images:", len(invalid_images))

print("\nDataset quality status:")

if len(missing_files) == 0:
    print("✅ No missing image files")

if len(zero_size_images) == 0:
    print("✅ No zero-size image files")

if len(invalid_images) == 0:
    print("✅ No invalid/corrupted images")

if missing_files or zero_size_images or invalid_images:
    print("\n⚠️ Some image quality problems were detected.")
else:
    print("\n✅ All checked images appear valid.")
```
