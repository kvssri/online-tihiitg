```python
# Day 5 - Pipeline & Experiment Planning

print("==============================================")
print("DAY 5 - PIPELINE & EXPERIMENT PLANNING")
print("==============================================")

print("\nObjectives:")
print("1. Design a reproducible workflow")
print("2. Define baseline and advanced methods")
print("3. Define project-specific experiment")
print("4. Define evaluation criteria")
print("5. Create experiment log")
print("6. Define repository and folder structure")

print("\nExpected Outputs:")
print("- Pipeline flowchart")
print("- Experiment plan")
print("- Experiment log template")
print("- Repository/folder structure")
```

```python
print("==============================================")
print("REPRODUCIBLE SOH / RUL PREDICTION PIPELINE")
print("==============================================")

pipeline = [
    "1. Raw Battery Data",
    "        ↓",
    "2. Data Validation & Quality Checks",
    "        ↓",
    "3. Leakage-Safe Train / Validation / Test Split",
    "        ↓",
    "4. Preprocessing & Normalization",
    "        ↓",
    "5. Sequence Construction",
    "        ↓",
    "6. Baseline Model",
    "        ↓",
    "7. Advanced Models (LSTM / GRU / TCN)",
    "        ↓",
    "8. SoH / RUL Prediction",
    "        ↓",
    "9. Evaluation",
    "        ↓",
    "10. Comparison & Final Results"
]

for step in pipeline:
    print(step)
```

```python
print("==============================================")
print("EXPERIMENT PLAN")
print("==============================================")

experiments = [
    ("Experiment 1", "Classical Baseline",
     "Linear / Polynomial Regression",
     "Establish a simple reference performance"),

    ("Experiment 2", "LSTM",
     "Long Short-Term Memory",
     "Learn temporal battery degradation patterns"),

    ("Experiment 3", "GRU",
     "Gated Recurrent Unit",
     "Compare performance with LSTM using the same data and metrics"),

    ("Experiment 4", "TCN",
     "Temporal Convolutional Network",
     "Evaluate an alternative temporal modelling approach"),

    ("Experiment 5", "Project-Specific Experiment",
     "Selected best-performing approach + project-specific preprocessing/sequence setup",
     "Evaluate the final proposed approach")
]

for exp, method, approach, objective in experiments:
    print(f"\n{exp}")
    print("Method   :", method)
    print("Approach :", approach)
    print("Objective:", objective)
```

```python
print("==============================================")
print("EVALUATION CRITERIA")
print("==============================================")

metrics = {
    "MAE": "Mean Absolute Error - average prediction error",
    "RMSE": "Root Mean Squared Error - gives more weight to large errors",
    "MAPE": "Mean Absolute Percentage Error - error expressed as percentage",
    "R2": "R-squared - measures how well predictions explain the target",
    "Training Time": "Time required to train the model",
    "Inference Time": "Time required to generate predictions",
    "Model Size": "Storage/memory required by the trained model"
}

for metric, meaning in metrics.items():
    print(f"\n{metric}")
    print("→", meaning)

print("\n==============================================")
print("COMPARISON RULE")
print("==============================================")

print("""
All models will be evaluated on the same test set.
The same target definition, preprocessing strategy,
and evaluation metrics will be used wherever applicable.
""")
```
