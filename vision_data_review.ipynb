```python
image_extensions = (
    '.jpg', '.jpeg', '.png',
    '.bmp', '.webp',
    '.tif', '.tiff'
)

image_files = []

for root, dirs, files in os.walk('/content/project_data'):

    for file in files:

        if file.lower().endswith(image_extensions):
            image_files.append(
                os.path.join(root, file)
            )

print("Total images found:", len(image_files))

print("\nFirst 10 images:")

for img in image_files[:10]:
    print(img)
```

```python
image_files = sorted(image_files)

print("Images sorted successfully!")

for img in image_files[:10]:
    print(img)
```

```python
SEQUENCE_LENGTH = 5

print("Sequence length:", SEQUENCE_LENGTH)
```

```python
from torchvision import transforms

IMAGE_SIZE = 256

preprocess = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.5, 0.5, 0.5],
        std=[0.5, 0.5, 0.5]
    )
])

print("Preprocessing pipeline created!")
```

```python
sequences = []

for i in range(
    len(image_files) - SEQUENCE_LENGTH + 1
):

    sequence = image_files[
        i:i + SEQUENCE_LENGTH
    ]

    sequences.append(sequence)

print("Total sequences:", len(sequences))

print("\nExample sequence:")

for img in sequences[0]:
    print(img)
```

```python
input_sequences = []
target_images = []

for sequence in sequences:

    input_sequences.append(
        sequence[:-1]
    )

    target_images.append(
        sequence[-1]
    )

print("Input sequences:", len(input_sequences))
print("Target images:", len(target_images))
```

```python
print("Example Input Sequence:")

for img in input_sequences[0]:
    print(img)

print("\nTarget Image:")
print(target_images[0])
```

```python
from sklearn.model_selection import train_test_split

SEED = 42

train_seq, temp_seq = train_test_split(
    sequences,
    test_size=0.30,
    random_state=SEED
)

val_seq, test_seq = train_test_split(
    temp_seq,
    test_size=0.50,
    random_state=SEED
)

print("Training sequences:", len(train_seq))
print("Validation sequences:", len(val_seq))
print("Testing sequences:", len(test_seq))
```

```python
def get_images(sequence_list):
    return set(
        img
        for seq in sequence_list
        for img in seq
    )

train_images = get_images(train_seq)
val_images = get_images(val_seq)
test_images = get_images(test_seq)

print(
    "Train-Val image overlap:",
    len(train_images & val_images)
)

print(
    "Train-Test image overlap:",
    len(train_images & test_images)
)

print(
    "Val-Test image overlap:",
    len(val_images & test_images)
)
```

```python
import pandas as pd

rows = []

for split_name, split_data in [
    ("train", train_seq),
    ("validation", val_seq),
    ("test", test_seq)
]:

    for i, seq in enumerate(split_data):

        rows.append({
            "split": split_name,
            "sequence_id": f"{split_name}_{i+1}",
            "sequence_length": len(seq),
            "input_images": len(seq) - 1,
            "target_image": seq[-1]
        })

sequence_df = pd.DataFrame(rows)

sequence_df.head()
```

```python
sequence_df.to_csv(
    "dataset_v1_sequences.csv",
    index=False
)

print("Sequence file saved successfully!")
```
