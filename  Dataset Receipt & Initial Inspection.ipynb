```python
import os
import shutil

zip_file = "Underwater-Image-Data-set-main.zip"

raw_folder = "raw_data"
os.makedirs(raw_folder, exist_ok=True)

raw_copy = os.path.join(raw_folder, zip_file)

shutil.copy2(zip_file, raw_copy)

print("Original dataset preserved.")
print("Raw archive:", raw_copy)
print("File size:", os.path.getsize(raw_copy), "bytes")
```

```python
import zipfile

with zipfile.ZipFile(zip_file, "r") as zip_ref:
    zip_contents = zip_ref.namelist()

print("Total items in ZIP:", len(zip_contents))

print("\nFirst 50 items:")
for item in zip_contents[:50]:
    print(item)
```

```python
from collections import Counter
import os

extensions = Counter()

for item in zip_contents:
    if not item.endswith("/"):
        ext = os.path.splitext(item)[1].lower()

        if ext:
            extensions[ext] += 1
        else:
            extensions["[no extension]"] += 1

print("===================================")
print("FILE TYPE SUMMARY")
print("===================================")

for ext, count in extensions.items():
    print(f"{ext}: {count}")

print("\nTotal files:", sum(extensions.values()))
```

```python
import os
import zipfile

inspection_folder = "inspection_data"

os.makedirs(inspection_folder, exist_ok=True)

with zipfile.ZipFile(zip_file, "r") as zip_ref:
    zip_ref.extractall(inspection_folder)

print("Dataset extracted for inspection.")
print("Inspection folder:", inspection_folder)
```

```python
image_extensions = [".jpg", ".jpeg", ".png", ".bmp", ".webp"]

image_files = []

for root, dirs, files in os.walk(inspection_folder):
    for file in files:
        if os.path.splitext(file)[1].lower() in image_extensions:
            image_files.append(os.path.join(root, file))

print("===================================")
print("IMAGE SUMMARY")
print("===================================")

print("Total images:", len(image_files))

print("\nFirst 10 images:")
for file in image_files[:10]:
    print("-", file)
```

```python
from PIL import Image
from collections import Counter

valid_images = 0
invalid_images = []
image_sizes = Counter()

for file in image_files:
    try:
        with Image.open(file) as img:
            img.verify()

        with Image.open(file) as img:
            image_sizes[img.size] += 1

        valid_images += 1

    except Exception:
        invalid_images.append(file)

print("===================================")
print("IMAGE QUALITY CHECK")
print("===================================")

print("Total images:", len(image_files))
print("Valid images:", valid_images)
print("Invalid/corrupted images:", len(invalid_images))

print("\nMost common dimensions:")
for size, count in image_sizes.most_common(10):
    print(size, ":", count)
```

```python
label_extensions = [
    ".csv",
    ".json",
    ".txt",
    ".xml",
    ".xlsx"
]

label_files = []

for root, dirs, files in os.walk(inspection_folder):
    for file in files:
        if os.path.splitext(file)[1].lower() in label_extensions:
            label_files.append(os.path.join(root, file))

print("===================================")
print("LABEL / METADATA FILES")
print("===================================")

if label_files:
    for file in label_files:
        print("📄", file)

    print("\nTotal possible label/metadata files:", len(label_files))
else:
    print("No common label/metadata files found.")
```

```python
import hashlib
from collections import defaultdict

hashes = defaultdict(list)

for file in image_files:
    try:
        with open(file, "rb") as f:
            file_hash = hashlib.md5(f.read()).hexdigest()

        hashes[file_hash].append(file)

    except Exception:
        pass

duplicate_groups = [
    files for files in hashes.values()
    if len(files) > 1
]

duplicate_count = sum(len(group) - 1 for group in duplicate_groups)

print("===================================")
print("DUPLICATE CHECK")
print("===================================")

print("Duplicate groups:", len(duplicate_groups))
print("Duplicate files:", duplicate_count)

if duplicate_groups:
    print("\nExamples:")
    for group in duplicate_groups[:5]:
        print(group)
else:
    print("✅ No exact duplicate images detected.")
```

```python
import matplotlib.pyplot as plt
from PIL import Image
import random

sample_count = min(12, len(image_files))

if sample_count > 0:
    samples = random.sample(image_files, sample_count)

    plt.figure(figsize=(15, 10))

    for i, file in enumerate(samples):
        img = Image.open(file)

        plt.subplot(3, 4, i + 1)
        plt.imshow(img)
        plt.axis("off")
        plt.title(os.path.basename(file)[:20])

    plt.tight_layout()
    plt.show()
else:
    print("No images available for visualization.")
```
