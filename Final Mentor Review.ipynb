# 2. Final Methodology Decisions

## Data Preparation
- Raw data is preserved without modification.
- Data quality and missing-value issues are inspected before preprocessing.
- Duplicate and inconsistent samples are checked.

## Data Splitting
A leakage-safe train/validation/test strategy is used.

Proposed split:
- Training: 70%
- Validation: 15%
- Testing: 15%

Splitting is performed carefully to avoid information from the test set entering training.

## Normalisation
Normalisation parameters are calculated using training data only.

## Sequence Construction
Historical observations are converted into fixed-length sequences for recurrent models.

## Modelling
The project compares a classical baseline with deep-learning approaches such as LSTM and GRU.
